## The Four Pillars of OOP

Object-Oriented Programming (OOP) is built upon four fundamental principles, often referred to as pillars:

### 1. Encapsulation

*   **Concept:** Bundling data (attributes) and methods (functions) that operate on the data into a single unit (a class). It also involves restricting direct access to some of an object's components, meaning internal workings are hidden, and interaction is through well-defined interfaces.
*   **Real-World Example:** A **smartphone**. You know how to use its buttons and touch screen (the public interface) to make calls, send messages, or launch apps. You don't need to know *how* the internal circuits, processor, or software work (the hidden implementation) to use it effectively. The phone's internal components and operations are encapsulated.
*   **Why it's useful:** Protects data integrity, makes code more modular and easier to maintain, and reduces complexity.

HIDING THE DATA with __ underscore underscore

DATA HIDING AND ACCESS CONTROL

### 2. Inheritance

*   **Concept:** Allows a new class (subclass/child class) to inherit properties and behaviors (attributes and methods) from an existing class (superclass/parent class). This promotes code reuse and establishes a "is-a" relationship.
*   **Real-World Example:** Different types of **vehicles**. A `Car` "is a" type of `Vehicle`, and a `Motorcycle` "is a" type of `Vehicle`. Both `Car` and `Motorcycle` inherit common characteristics from `Vehicle` (e.g., they both have wheels, an engine, and can move) but also have their unique features (a car has 4 wheels, a motorcycle has 2).
*   **Why it's useful:** Promotes code reusability, reduces redundancy, and allows for building specialized classes from more general ones.

### 3. Polymorphism

*   **Concept:** "Many forms." It allows objects of different classes to be treated as objects of a common superclass. This means a single interface can be used for different data types or classes.
*   **Real-World Example:** A **remote control**. You can use one remote control (the interface) to operate different devices (objects) like a TV, a DVD player, or a stereo. Pressing the "Play" button performs a "play" action, but the specific way the TV "plays" (shows video) is different from how the stereo "plays" (produces audio). The common interface ("Play" button) has different implementations based on the object.
*   **Why it's useful:** Provides flexibility, makes code more generic and extensible, and allows for cleaner, more readable code.

### 4. Abstraction

*   **Concept:** Hiding complex implementation details and showing only the essential features of an object. It focuses on "what it does" rather than "how it does it."
*   **Real-World Example:** Driving a **car**. When you drive, you interact with the steering wheel, pedals, and gear shift. You know *what* these controls do (accelerate, brake, turn) without needing to understand the intricate mechanics of the engine, transmission, or braking system *how* they achieve those actions. The complex inner workings are abstracted away.
*   **Why it's useful:** Manages complexity, improves code readability, and makes it easier to change internal implementations without affecting external usage.

***getters*** and ***setters*** (also known as accessors and mutators) are public methods used to access and modify a class's private fields. They are the primary mechanism for implementing encapsulation, ensuring that internal data is protected from unauthorized or invalid changes.

````python
#Encapsulation code example
class Student:
    def __init__(self, name, grade):
        self.name = name          # Public attribute
        self.__grade = grade      # Encapsulated (private-like) attribute

    def get_grade(self):
        return self.__grade # Public method to *read* the grade

    def set_grade(self, new_grade):
        # Public method to *write* the grade (can add validation here if needed)
        if new_grade >= 0:
            self.__grade = new_grade
        else:
            print("Grade cannot be negative!")

# --- How it works ---
alice = Student("Shiv", 85)

print(f"Shiv's name: {alice.name}")
print(f"Shiv's grade (via getter): {alice.get_grade()}")

alice.set_grade(90) # Change grade via setter
print(f"Shiv's new grade: {alice.get_grade()}")

alice.set_grade(-5) # Attempt to set invalid grade
print(f"Shiv's grade after invalid attempt: {alice.get_grade()}")

# Direct access to encapsulated attribute is discouraged
# print(alice.__grade) # This would typically cause an AttributeError
````

In the above example, __grade is our 'private' attribute. You can only interact
with it using the get_grade() and set_grade() methods, which gives you control

````python
# class account
# - constructor(self and balance)
# - get_balance (pass self)
# - deposit (self,amount)
# - withdraw (self,amount)Take input as function and give output as a function

class Account:
    def __init__(self, initial_balance=0):
        """Constructor to initialize the account with an initial balance.
        The balance is encapsulated (hidden) using __balance.
        It ensures the initial balance is not negative.
        """
        if initial_balance >= 0:
            self.__balance = initial_balance
        else:
            self.__balance = 0
            print("Initial balance cannot be negative. Setting to 0.")

    def get_balance(self):
        """Getter method to safely retrieve the current account balance."""
        return self.__balance

    def deposit(self, amount):
        """Deposits a specified amount into the account.
        Only positive amounts can be deposited.
        """
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount:.2f}. New balance: ${self.__balance:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        """Withdraws a specified amount from the account.
        Checks for positive withdrawal amount and sufficient funds.
        """
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self.__balance:
            print("Insufficient funds. Current balance: ${:.2f}.".format(self.__balance))
        else:
            self.__balance -= amount
            print(f"Withdrew ${amount:.2f}. New balance: ${self.__balance:.2f}")

# --- Usage Examples and Explanation of What's Done ---

# 1. Creating an account object:
# An instance of the Account class is created, initiating the constructor (__init__).
# The initial_balance is set to 1000. If a negative balance was passed, it would default to 0.
my_account = Account(1000)
print(f"Initial Balance (via get_balance()): ${my_account.get_balance():.2f}")

# 2. Performing a deposit:
# The deposit method is called, adding 500 to the account balance.
my_account.deposit(500)

# 3. Performing a withdrawal:
# The withdraw method is called, subtracting 200 from the balance.
my_account.withdraw(200)

# 4. Another deposit with a decimal amount:
# Demonstrates handling of floating-point numbers for transactions.
my_account.deposit(150.75)

# 5. Attempting an insufficient withdrawal:
# This call to withdraw will be rejected because the amount (1500) exceeds the current balance.
my_account.withdraw(1500)

# 6. Attempting a negative withdrawal:
# This call to withdraw will be rejected because the amount is not positive.
my_account.withdraw(-50)

# 7. Getting the final balance:
# The get_balance method is used again to show the account's state after all operations.
print(f"Final Balance (via get_balance()): ${my_account.get_balance():.2f}")
````

````python
import abc # Import the 'abc' module for Abstract Base Classes

# Define an Abstract Base Class (ABC) named 'Shape'
# To make a class an ABC, it must inherit from abc.ABC
class Shape(abc.ABC):
    # This is an abstract method. It declares an interface but provides no implementation.
    # Any concrete class inheriting from Shape *must* provide an implementation for 'area'.
    @abc.abstractmethod
    def area(self):
        pass # No implementation here, just a declaration

    # This is also an abstract method.
    @abc.abstractmethod
    def perimeter(self):
        pass # No implementation here, just a declaration

    # This is a concrete method. It has an implementation and can be inherited directly.
    def describe(self):
        print("This is a geometric shape.")

# Define a concrete class 'Circle' that inherits from 'Shape'
# Circle must implement all abstract methods from Shape.
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    # Implement the 'area' method for Circle
    def area(self):
        return 3.14 * self.radius * self.radius

    # Implement the 'perimeter' method for Circle
    def perimeter(self):
        return 2 * 3.14 * self.radius

# Define another concrete class 'Rectangle' that inherits from 'Shape'
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    # Implement the 'area' method for Rectangle
    def area(self):
        return self.width * self.height

    # Implement the 'perimeter' method for Rectangle
    def perimeter(self):
        return 2 * (self.width + self.height)

# --- Usage and Demonstration of Abstraction ---

# Create instances of the concrete classes
my_circle = Circle(5)
my_rectangle = Rectangle(4, 6)

# You cannot directly instantiate an abstract class:
# try:
#     abstract_shape = Shape()
# except TypeError as e:
#     print(f"Cannot instantiate abstract class: {e}")

print(f"Circle Area: {my_circle.area()}")
print(f"Circle Perimeter: {my_circle.perimeter()}")
my_circle.describe() # Inherited concrete method

print(f"\nRectangle Area: {my_rectangle.area()}")
print(f"Rectangle Perimeter: {my_rectangle.perimeter()}")
my_rectangle.describe() # Inherited concrete method

# Abstraction allows us to treat different shapes polymorphically
shapes = [my_circle, my_rectangle]

print("\n--- Processing shapes abstractly ---")
for shape in shapes:
    shape.describe()
    print(f"  Area: {shape.area()}")
    print(f"  Perimeter: {shape.perimeter()}")
````

## Types of Inheritance Explained in Simple Terms

Inheritance is a fundamental concept in Object-Oriented Programming (OOP) where a new class (child/subclass) derives properties and behaviors (attributes and methods) from an existing class (parent/superclass). This promotes code reusability and establishes an 'is-a' relationship.

### 1. Single Inheritance

*   **Concept:** A child class inherits from only one parent class.
*   **Simple Analogy:** A child inherits traits from *one* parent.
*   **Example:** A `Dog` is a type of `Animal`. The `Dog` class inherits from the `Animal` class.

````python
# Single Inheritance Example
class Animal:
    def speak(self):
        print("Animal makes a sound.")

class Dog(Animal): # Dog inherits from Animal
    def bark(self):
        print("Dog barks.")

my_dog = Dog()
my_dog.speak() # Inherited from Animal
my_dog.bark()  # Specific to Dog
````

### 2. Multiple Inheritance

*   **Concept:** A child class inherits from multiple parent classes.
*   **Simple Analogy:** A child inherits traits from *two or more* parents (e.g., father and mother).
*   **Example:** A `StudentAssistant` might inherit qualities from both a `Student` and a `Teacher`.

````python
# Multiple Inheritance Example
class Father:
    def farming_skills(self):
        print("Father has farming skills.")

class Mother:
    def cooking_skills(self):
        print("Mother has cooking skills.")

class Child(Father, Mother): # Child inherits from both Father and Mother
    def unique_talent(self):
        print("Child has a unique talent.")

my_child = Child()
my_child.farming_skills()
my_child.cooking_skills()
my_child.unique_talent()
````

### 3. Multilevel Inheritance

*   **Concept:** A child class inherits from a parent class, which in turn inherits from another parent class. It forms a chain of inheritance.
*   **Simple Analogy:** Grandparent -> Parent -> Child. Traits are passed down through generations.
*   **Example:** A `Car` is a type of `Vehicle`, and a `SportsCar` is a type of `Car`.

````python
# Multilevel Inheritance Example
class Grandparent:
    def gene_A(self):
        print("Has gene A.")

class Parent(Grandparent): # Parent inherits from Grandparent
    def gene_B(self):
        print("Has gene B.")

class Child(Parent): # Child inherits from Parent (and indirectly from Grandparent)
    def gene_C(self):
        print("Has gene C.")

my_family_child = Child()
my_family_child.gene_A() # Inherited from Grandparent
my_family_child.gene_B() # Inherited from Parent
my_family_child.gene_C() # Specific to Child
````

### 4. Hierarchical Inheritance

*   **Concept:** Multiple child classes inherit from a single parent class.
*   **Simple Analogy:** One parent has multiple children, and each child is distinct.
*   **Example:** An `Animal` class could have `Dog`, `Cat`, and `Cow` classes inheriting from it.

````python
# Hierarchical Inheritance Example
class Shape:
    def draw(self):
        print("Drawing a shape.")

class Circle(Shape): # Circle inherits from Shape
    def draw_circle(self):
        print("Drawing a circle.")

class Square(Shape): # Square also inherits from Shape
    def draw_square(self):
        print("Drawing a square.")

my_circle = Circle()
my_circle.draw() # Inherited
my_circle.draw_circle()

my_square = Square()
my_square.draw() # Inherited
my_square.draw_square()
````

### 5. Hybrid Inheritance (Combination)

*   **Concept:** This is a combination of two or more types of inheritance (e.g., a mix of multiple and multilevel inheritance).
*   **Simple Analogy:** A complex family tree with various branches and interconnections.
*   **Example:** Imagine a `Classroom` system where `Teacher` and `Student` are types of `Person`, and `TeachingAssistant` inherits from both `Teacher` and `Student`.

````python
# Hybrid Inheritance Example (Combination of Hierarchical and Multiple)
class Person:
    def greet(self):
        print("Hello!")

class Teacher(Person):
    def teach(self):
        print("Teaching.")

class Student(Person):
    def learn(self):
        print("Learning.")

class TeachingAssistant(Teacher, Student): # Multiple Inheritance
    def assist(self):
        print("Assisting both teaching and learning.")

ta = TeachingAssistant()
ta.greet() # Inherited from Person (via Teacher or Student)
ta.teach() # Inherited from Teacher
ta.learn() # Inherited from Student
ta.assist() # Specific to TeachingAssistant
````

### Public Members (Accessible by Everyone)

*   **Concept:** Think of public members as the parts of a car that are visible and directly usable by the driver and passengers. Anyone can interact with them.
*   **Python Convention:** In Python, any variable or function (method) inside a class that doesn't start with an underscore (`_`) is considered public. There's no special keyword for it; it's the default.
*   **Real-World Example:** The **steering wheel**, **pedals (accelerator, brake)**, **gear shift**, and **radio buttons** in a car. You, as the user, directly interact with these to control the car. They are meant to be used by anyone operating the vehicle.

### Protected Members (Accessible by the Class and its Family)

*   **Concept:** Protected members are like parts of a car that are meant for the mechanic or the car manufacturer to use or modify, not typically the driver. They are *internal* but can be accessed by special 'family members' (subclasses).
*   **Python Convention:** A single leading underscore (`_`) signifies a protected member (e.g., `_internal_setting`). Python doesn't strictly *prevent* you from accessing these from outside, but it's a strong hint: "hands off, unless you know what you're doing, or you're a related class."
*   **Real-World Example:** The **engine oil dipstick** or the **fuse box** in a car. A regular driver might know where they are, but typically only a mechanic or someone knowledgeable about the car's maintenance would interact with them directly. They are internal components but can be accessed for maintenance or adjustments by those who 'inherit' the knowledge or tools.

### Private Members (Strictly for Internal Use of the Class)

*   **Concept:** Private members are like the intricate internal mechanisms of a car's engine or transmission. They are crucial for the car to function, but a driver never interacts with them directly, and even mechanics usually don't mess with them unless performing a deep overhaul. They are hidden and guarded.
*   **Python Convention:** A double leading underscore (`__`) signifies a private member (e.g., `__hidden_logic`). Python actively renames these members (this is called 'name mangling') to make them much harder to access from outside the class or from subclasses. This helps prevent accidental modification and ensures that the internal workings can change without affecting other parts of your code.
*   **Real-World Example:** The **specific programming code in the car's onboard computer** that manages fuel injection timing or automatic transmission shifting. These are highly internal details. A driver doesn't even know they exist, and even a mechanic typically uses diagnostic tools (public methods) to interact with the computer, rather than directly altering its internal code. The manufacturer carefully designs these to be internal and self-contained.

## The Diamond Problem: A Beginner's Guide

Imagine you're building a family tree for software classes using inheritance. Inheritance is when one class (child) gets features from another class (parent).

### What is Multiple Inheritance?

Normally, a class inherits from one parent. But in some programming languages (like Python), a class can inherit from *multiple* parents. This is called **Multiple Inheritance**.

### The Problem: When Paths Cross

The 'Diamond Problem' occurs when you have a specific setup with multiple inheritance:

1.  **Class A** is a base class (the 'grandparent').
2.  **Class B** inherits from A.
3.  **Class C** also inherits from A.
4.  **Class D** inherits from *both* B and C (the 'child' at the bottom of the diamond).

If Class A has a method (let's say `do_something()`), and both Class B and Class C *change* or *override* that method in their own ways, what happens when Class D tries to call `do_something()`? Does it use B's version or C's version? This ambiguity is the diamond problem.

### Visualize the Diamond

Imagine a diagram:

```
      A
     / \
    B   C
     \ /
      D
```

*   `A` is at the top.
*   `B` and `C` are in the middle, each inheriting from `A`.
*   `D` is at the bottom, inheriting from both `B` and `C`.

If `A` has `do_something()`, and `B` and `C` each have their own `do_something()`, then `D` has *two* possible paths to `do_something()` through `B` and `C`, both of which ultimately originate from `A`. This creates confusion: which `do_something()` should `D` use?

### Real-World Example: The "Smart Gadget"

Let's use a real-world analogy:

*   **Device (A):** A basic electronic device with a method `turn_on()`. (This just prints "Device turning on")

*   **Speaker (B):** Inherits from `Device`. It overrides `turn_on()` to also play a startup sound. (`super().turn_on()` then `play_sound()`).

*   **Camera (C):** Also inherits from `Device`. It overrides `turn_on()` to also initialize sensors. (`super().turn_on()` then `init_sensors()`).

*   **SmartPhone (D):** This is your new phone. It inherits from *both* `Speaker` and `Camera`.

When you `smartphone.turn_on()`, what should happen? Should it play the sound first, then initialize sensors? Or initialize sensors first, then play the sound? Both `Speaker` and `Camera` are trying to modify the `turn_on()` behavior inherited from `Device`. This is the diamond problem in action.

### How Python Solves It: Method Resolution Order (MRO)

Python handles the diamond problem using something called the **Method Resolution Order (MRO)**. The MRO is a specific linear order in which Python searches for methods and attributes in a class hierarchy, especially with multiple inheritance. Python uses an algorithm called C3 linearization to determine this order, which ensures consistency and predictability.

You can actually see the MRO for any class using `ClassName.mro()` or `ClassName.__mro__`.

When `D` calls `do_something()`, Python looks at its MRO list. It goes through the parent classes in that specific order, and the first `do_something()` method it finds in that order is the one it executes. This removes the ambiguity.


````python
# Python demonstration of the Diamond Problem and MRO

class Device: # Class A
    def turn_on(self):
        print("Device: Turning on.")

class Speaker(Device): # Class B inherits from A
    def turn_on(self):
        super().turn_on() # Call parent's turn_on
        print("Speaker: Playing startup sound.")

class Camera(Device): # Class C inherits from A
    def turn_on(self):
        super().turn_on() # Call parent's turn_on
        print("Camera: Initializing sensors.")

class SmartPhone(Speaker, Camera): # Class D inherits from B and C
    def turn_on(self):
        # Without super() here, it would only call the method from the first parent (Speaker)
        # With super(), it will follow the MRO and call the next method in line.
        super().turn_on()
        print("SmartPhone: Booting up applications.")

print("--- Diamond Problem Demonstration ---")
my_phone = SmartPhone()
my_phone.turn_on()

print("\n--- Understanding the MRO ---")
print("Method Resolution Order for SmartPhone:")
for cls in SmartPhone.__mro__:
    print(cls.__name__)

print("\nNotice how 'Device' is called only once, and the order 'Speaker' then 'Camera' is maintained before the final 'SmartPhone' action. Python's MRO resolves the ambiguity by providing a clear, consistent search path for methods.")
````

## Polymorphism: Many Forms, One Interface

*   **Concept:** Polymorphism literally means "many forms." In programming, it refers to the ability of different objects to respond to the same method call (or function) in their own specific ways. It allows you to use a common interface for different data types or classes.

*   **Simple Idea:** Imagine you have a general command, but how that command is executed depends on *who* or *what* receives it. The same command behaves differently based on the object it's applied to.

### Real-World Example: "Make Sound"

Think about the command "**Make Sound**":

*   If you tell a **Dog** to "Make Sound", it will **bark**.
*   If you tell a **Cat** to "Make Sound", it will **meow**.
*   If you tell a **Cow** to "Make Sound", it will **moo**.

Here, "Make Sound" is the single interface (the common action), but the actual behavior (bark, meow, moo) is different for each animal. This is polymorphism in action!

````python
class Dog:
  def speak(self):
    print("Woof!")

class Cat:
  def speak(self):
    print("Meow!")
d1=Dog()
d1.speak()
c1=Cat()
c1.speak()
````

````python
# Basic Code Example of Polymorphism

class Dog:
    def make_sound(self):
        print("Woof! Woof!")

class Cat:
    def make_sound(self):
        print("Meow!")

class Cow:
    def make_sound(self):
        print("Moooo!")

# A function that takes an animal object and tells it to make a sound
def animal_makes_a_sound(animal):
    animal.make_sound()

# Create instances of different animals
my_dog = Dog()
my_cat = Cat()
my_cow = Cow()

print("--- Demonstrating Polymorphism ---")

animal_makes_a_sound(my_dog)
animal_makes_a_sound(my_cat)
animal_makes_a_sound(my_cow)

print("\nNotice how the same function call 'animal_makes_a_sound()' behaves differently depending on the type of animal object it receives. This is polymorphism!")
````

### Abstraction: Showing Essential Features, Hiding Complexity

*   **Concept:** Abstraction is about hiding complex implementation details and showing only the essential features of an object. It focuses on "what it does" rather than "how it does it."

*   **Real-World Example:** Driving a **car**. When you drive, you interact with the steering wheel, pedals, and gear shift. You know *what* these controls do (accelerate, brake, turn) without needing to understand the intricate mechanics of the engine, transmission, or braking system *how* they achieve those actions. The complex inner workings are abstracted away.


````python
import abc # Import the 'abc' module for Abstract Base Classes

class Vehicle(abc.ABC): # Our abstract base class
    # This is an abstract method. It defines *what* a vehicle can do (start the engine)
    # but doesn't provide the *how*. Subclasses *must* implement this.
    @abc.abstractmethod
    def start_engine(self):
        pass

    # This is a concrete method, showing essential features.
    # It uses the abstract method but doesn't reveal its internal workings.
    def drive(self):
        self.start_engine() # Calls the abstract method, whose implementation is hidden
        print("Vehicle is moving.")

class Car(Vehicle): # A concrete subclass of Vehicle
    def start_engine(self):
        # This is the concrete implementation of 'start_engine' for a Car.
        # The complex details (e.g., fuel injection, spark plugs) are hidden here.
        print("Car engine starts with a turn of the key/push of a button.")

class Bicycle(Vehicle): # Another concrete subclass
    def start_engine(self):
        # For a bicycle, 'starting the engine' means something different (pedaling).
        # The specific implementation is hidden within this class.
        print("Bicycle starts with pedaling power.")

# --- Demonstration ---
# We can interact with different vehicles using a common 'drive' interface,
# without knowing the specific 'start_engine' mechanism of each.

print("--- Car Abstraction ---")
my_car = Car()
my_car.drive() # We just call 'drive', and the car handles its internal start_engine logic

print("\n--- Bicycle Abstraction ---")
my_bicycle = Bicycle()
my_bicycle.drive() # We just call 'drive', and the bicycle handles its internal start_engine logic

# You cannot create an object directly from an abstract class (like Vehicle)
# because it has unimplemented abstract methods. It only defines the interface.
# try:
#     abstract_vehicle = Vehicle()
# except TypeError as e:
#     print(f"\nError: {e}")
````

## Composition: "Has-A" Relationship

**Composition** is a fundamental concept in Object-Oriented Programming (OOP) that describes a "has-a" relationship between objects. Instead of inheriting properties from another class (which is an "is-a" relationship, like a `Dog` is an `Animal`), composition means one class *contains* or *uses* an object of another class as part of its own structure or functionality.

Think of it as building complex objects by putting together simpler, independent parts.

### Simple Analogy: A Car

Imagine a `Car`.

*   A `Car` **has a** `Engine`.
*   A `Car` **has a** `Wheel` (multiple, actually).
*   A `Car` **has a** `SteeringWheel`.
*   A `Car` **has a** `GPS` system.

In this scenario, the `Car` doesn't *inherit* from `Engine` or `Wheel`. A car *is not* an engine; it *has* an engine. The `Engine`, `Wheel`, `SteeringWheel`, and `GPS` are independent objects that work together to make the `Car` function. The `Car` class *composes* itself using these other objects.

### Why use Composition?

*   **Flexibility:** You can easily swap out components. If you want a different type of engine, you just replace the `Engine` object within the `Car`, without changing the core `Car` class structure.
*   **Reusability:** The `Engine` class can be reused in other types of vehicles (like a `Truck` or a `Boat`) without needing the `Truck` or `Boat` classes to inherit from it.
*   **Loose Coupling:** Changes in one component (e.g., how the `Engine` works internally) are less likely to break other components, as long as the way the `Car` interacts with the `Engine` (its public methods) remains consistent.
*   **Simplicity:** It helps manage complexity by breaking down a large problem into smaller, more manageable pieces.

````python
# Python Example of Composition: A Car and its parts

class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"Engine ({self.horsepower}hp) starting... Vroom!"

    def stop(self):
        return "Engine stopping."

class Wheel:
    def __init__(self, size):
        self.size = size

    def rotate(self):
        return f"Wheel ({self.size} inches) rotating."

class GPS:
    def __init__(self, provider):
        self.provider = provider

    def navigate(self, destination):
        return f"GPS ({self.provider}) navigating to {destination}."

class Car:
    def __init__(self, make, model, engine_horsepower, wheel_size, gps_provider):
        self.make = make
        self.model = model
        # Composition: A Car HAS-A Engine, HAS-A Wheel (for simplicity, one here), HAS-A GPS
        self.engine = Engine(engine_horsepower) # Car contains an Engine object
        self.front_left_wheel = Wheel(wheel_size) # Car contains a Wheel object
        self.navigation = GPS(gps_provider) # Car contains a GPS object

    def drive(self, destination):
        print(f"\nDriving the {self.make} {self.model}...")
        print(self.engine.start()) # Car uses its Engine's functionality
        print(self.front_left_wheel.rotate()) # Car uses its Wheel's functionality
        print(self.navigation.navigate(destination)) # Car uses its GPS's functionality
        print(f"{self.make} {self.model} arrived at {destination}.")
        print(self.engine.stop())

# --- Usage ---
my_car = Car("Toyota", "Camry", 180, 17, "Google Maps")
my_car.drive("the grocery store")

# Another car with different components
sports_car = Car("Ferrari", "488 GTB", 660, 20, "TomTom")
sports_car.drive("the race track")

# Notice how the Car class itself doesn't need to know the internal workings of
# Engine, Wheel, or GPS. It just knows *what* they can do (their public methods)
# and uses them to perform its own actions (driving). This is the essence of composition.
````

````python
#composition
class Adress:
  def __init__(self,city):
    self.city=city
  def show_address(self):
    print("City", self.city)
#student class
class Student:
  def __init__(self,name,city):
    self.name=name
    #composition
    #creating object of adress ckass inside student class
    self.address=Adress(city)
  def show_student(self):
    print("NAME:",self.name)
    #using object of another class
    self.address.show_address()

s=Student("karan", "Delhi")
s.show_student()
````

````python
#create a class engine then create a class car
#car has a engine (do composition)
#in engine create one method name def start no constrcutor nothing and that is printing engine started
#in the car make a constructor one value is passed in the cell and we have to call the engine self.engine=engine()
#2nd method will be def drive in the car but before car is moving you have to print engine started

class Engine:
  # No constructor as per the requirement
  def start(self):
    print("Engine started")

class Car:
  def __init__(self, model): # One value passed to the constructor (e.g., car model)
    self.model = model
    # Composition: creating object of Engine class inside Car class
    self.engine = Engine()

  def drive(self):
    # First, the engine starts
    self.engine.start()
    # Then, the car is moving
    print(f"Car {self.model} is moving.")

my_car = Car("Sedan") # Pass a single value for the car (like 'Karan')
my_car.drive()

# print("\n--- Another Car ---")
# sports_car = Car("SportsCar") # Another instance
# sports_car.drive()
````

````python
#create a teacher and subject classes where teacher HAS A subject (has means inheritance)

class Subject:
  # No constructor as per the requirement (like Engine)
  def teach_topic(self):
    print("Subject: Learning Python OOP concepts.")

class Teacher:
  def __init__(self, name): # One value passed to the constructor (like Car's model)
    self.name = name
    # Composition: A Teacher HAS-A Subject
    # creating object of Subject class inside Teacher class
    self.subject = Subject()

  def teach(self):
    # First, the subject's topic is taught (like engine starts)
    self.subject.teach_topic()
    # Then, the teacher performs their action (like car moving)
    print(f"Teacher {self.name} is now conducting the class.")

# --- Usage (similar to the Car/Engine example) ---
# Create an instance of the Teacher class
prof_alex = Teacher("Professor Alex")
# Call the teach method
prof_alex.teach()

print("\n--- Another Teacher ---")
dr_smith = Teacher("Dr. Smith")
dr_smith.teach()
````

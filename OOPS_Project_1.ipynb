```python
#BANK PROJECT DECODED
class BankAccount: # Defines the base class for all bank accounts
    def __init__(self, acc_no, name, balance): # Constructor to initialize account details
        self.acc_no = acc_no # Account number
        self.name = name # Account holder's name
        self._balance = balance # Initial balance, stored as _balance

    def deposit(self, amount): # Method to deposit money into the account
        self._balance += amount # Adds the amount to the current _balance
        return self._balance # Returns the new _balance

    def withdraw(self, amount): # Method to withdraw money from the account
        if amount <= self._balance: # Checks if sufficient _balance is available
            self._balance -= amount # Deducts the amount from the _balance
            return self._balance # Returns the new _balance
        else:
            return "insufficient balance" # Returns a message if _balance is insufficient

    def get_balance(self): # Method to get the current account _balance
        return self._balance # Returns the current _balance

    def calculate_interest(self): # Placeholder method for calculating interest
        pass # To be implemented by subclasses


class SavingAccount(BankAccount): # Defines a subclass for savings accounts, inheriting from BankAccount
    def calculate_interest(self): # Overrides the calculate_interest method for savings accounts
        return self._balance * 0.04 # Calculates 4% interest based on _balance


class CurrentAccount(BankAccount): # Defines a subclass for current accounts, inheriting from BankAccount
    def calculate_interest(self): # Overrides the calculate_interest method for current accounts
        return 0 # Current accounts typically do not earn interest


class Bankapp: # Main application class to manage bank operations
    def __init__(self): # Constructor for the Bankapp
        self.account = None # Initializes account to None, will hold an account object

    def create_account(self, acc_no, name, balance, acc_type): # Method to create a new bank account
        if acc_type == "savings": # Checks if the account type is savings
            self.account = SavingAccount(acc_no, name, balance) # Creates a SavingAccount object
        else: # Otherwise, assumes it's a current account
            self.account = CurrentAccount(acc_no, name, balance) # Creates a CurrentAccount object
        return "account created successfully" # Returns a success message

    def deposit_money(self, amount): # Method to deposit money into the currently selected account
        return self.account.deposit(amount) # Calls the deposit method of the account object

    def withdraw_money(self, amount): # Method to withdraw money from the currently selected account
        return self.account.withdraw(amount) # Calls the withdraw method of the account object

    def check_balance(self): # Method to check the balance of the currently selected account
        return self.account.get_balance() # Calls the get_balance method of the account object

    def get_intrest(self): # Method to calculate interest for the currently selected account
        return self.account.calculate_interest() # Calls the calculate_interest method of the account object

app=Bankapp() # Creates an instance of the Bankapp
print(app.create_account(101,"pranit",10000,"savings")) # Creates a savings account and prints a confirmation
print("balance after desposit:" ,app.deposit_money(2000)) # Deposits money and prints the new balance
print("balance after withdraw:",app.withdraw_money(2000)) # Withdraws money and prints the new balance
print("current balance:",app.check_balance()) # Prints the current balance
print("intrest earned:",app.get_intrest()) # Prints the interest earned
```

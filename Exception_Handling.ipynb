### Understanding Exceptions and Exception Handling

An **exception** is an event that occurs during the execution of a program that disrupts the normal flow of instructions. When a Python script encounters a situation that it cannot cope with, it raises an exception. If unhandled, this will typically lead to the termination of the program.

**Exception handling** is a programming construct designed to handle the occurrence of exceptions, allowing the program to continue executing instead of crashing. This is achieved using `try`, `except`, `else`, and `finally` blocks:

*   **`try` block**: Contains the code segment where an exception might occur. If an exception happens here, the control is immediately passed to the corresponding `except` block.
*   **`except` block**: This block is executed if a specific type of exception (or any exception, if not specified) occurs within the `try` block. It contains the code to handle the error gracefully.
*   **`else` block**: (Optional) If present, the code inside the `else` block is executed only if the `try` block completes without raising an exception.
*   **`finally` block**: (Optional) This block always executes, regardless of whether an exception occurred or was handled. It's typically used for cleanup operations, such as closing files or releasing resources.

### Understanding ZeroDivisionError

A `ZeroDivisionError` is an arithmetic error that occurs when a program attempts to divide any number by zero. This operation is undefined in mathematics and consequently raises an error in Python.

When you divide any number with 0 --> print(10/0)

```python
# Example of a ZeroDivisionError

try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print(result)
except ZeroDivisionError as e:
    print(f"Caught a ZeroDivisionError: {e}")

# Another example
try:
    print(5 / (2 - 2))
except ZeroDivisionError as e:
    print(f"Caught another ZeroDivisionError: {e}")

# Correct division
print(f"Correct division: {10 / 2}")
```

### Understanding ValueError

A `ValueError` occurs when an operation or function receives an argument of the correct type but an inappropriate value. This often happens when a value is outside the expected range or format that a function can handle.

ValueError: When you give wrong value to a function --> num=int("abc")

```python
# Example of a ValueError

# Trying to convert a non-numeric string to an integer
try:
    invalid_number_str = "hello"
    num = int(invalid_number_str)
    print(num)
except ValueError as e:
    print(f"Caught a ValueError: {e}")

# Trying to find an element that does not exist in a list using .index()
try:
    my_list = [1, 2, 3]
    index = my_list.index(4)
    print(f"Index of 4: {index}")
except ValueError as e:
    print(f"Caught another ValueError: {e}")

# Correct operation
valid_number_str = "123"
num = int(valid_number_str)
print(f"Successfully converted '{valid_number_str}' to integer: {num}")
```

### Understanding TypeError

A `TypeError` is a common runtime error in Python that occurs when an operation or function is applied to an object of an inappropriate type. This means you're trying to do something with a variable that its data type doesn't support.

TypeError: When u perform an operation between different data types --> print(5+"abc)

```python
# Example of a TypeError

# Attempting to add an integer to a string
try:
    result = 5 + "hello"
    print(result)
except TypeError as e:
    print(f"Caught a TypeError: {e}")

# Attempting to call an integer as a function
try:
    x = 10
    x()
except TypeError as e:
    print(f"Caught another TypeError: {e}")

# Correct operations
print(f"Correct string concatenation: {'hello' + ' world'}")
print(f"Correct integer addition: {5 + 10}")
```

### Understanding IndexError

An `IndexError` is a common type of runtime error in Python. It occurs when you try to access an element in a sequence (like a list, tuple, or string) using an index that does not exist. This can happen if the index is negative and out of range, or if it's positive and greater than or equal to the length of the sequence.

When you access a wrong index --> lst[10,20,30]....print(lst[5])

```python
# Example of an IndexError
my_list = [10, 20, 30]

try:
    # This will cause an IndexError because index 3 is out of bounds
    print(my_list[3])
except IndexError as e:
    print(f"Caught an IndexError: {e}")

# This will also cause an IndexError for a negative index that's out of bounds
try:
    print(my_list[-4])
except IndexError as e:
    print(f"Caught another IndexError: {e}")

# Correct access
print(f"Accessing element at index 0: {my_list[0]}")
print(f"Accessing element at last index (-1): {my_list[-1]}")
```

### Understanding KeyError

A `KeyError` is a common type of runtime error in Python that occurs when you try to access a key that does not exist in a dictionary (or other mapping types). Dictionaries store data in key-value pairs, and if you attempt to retrieve a value using a key that is not present in the dictionary, a `KeyError` will be raised.

Keyerror: when u access a missing key in a dictionary -->

students={"name":"Rahul", "age":20}

print(students["city"])

```python
# Example of a KeyError
my_dict = {"name": "Alice", "age": 30, "city": "New York"}

try:
    # This will cause a KeyError because 'country' is not a key in my_dict
    print(my_dict["country"])
except KeyError as e:
    print(f"Caught a KeyError: {e}")

# Another example: using .get() method to avoid KeyError
# .get() returns None (or a default value if specified) instead of raising an error
country = my_dict.get("country", "Unknown")
print(f"Accessing 'country' with .get(): {country}")

# Correct access
print(f"Accessing 'name': {my_dict['name']}")
```

### Understanding FileNotFoundError

A `FileNotFoundError` is an `OSError` subclass that occurs when an attempt to access a file or directory fails because the specified file or directory does not exist. This is a common error when dealing with file operations, such as reading from a file or writing to a file.

when python not able to find file in ur pc

```python
# Example of a FileNotFoundError

try:
    # Attempting to open a file that does not exist
    with open('non_existent_file.txt', 'r') as f:
        content = f.read()
        print(content)
except FileNotFoundError as e:
    print(f"Caught a FileNotFoundError: {e}")

# Example of handling file creation
file_name = 'my_new_file.txt'
try:
    with open(file_name, 'r') as f:
        content = f.read()
        print(f"Content of '{file_name}':\n{content}")
except FileNotFoundError:
    print(f"File '{file_name}' not found. Creating it now.")
    with open(file_name, 'w') as f:
        f.write("This is a newly created file.")
    print(f"File '{file_name}' created successfully.")

# Verify content after creation
try:
    with open(file_name, 'r') as f:
        content = f.read()
        print(f"Content of '{file_name}' after creation:\n{content}")
except FileNotFoundError as e:
    print(f"Error verifying file: {e}")
```

```python
# CUSTOM ERRORS
class LowBalanceError(Exception):
  # This line creates a custom exception named LowBalanceError
  # pass is used because class cannot be empty in python
  pass

balance = 500
withdraw = int(input("Enter amount to withdraw: "))

try:
  if withdraw > balance:
    raise LowBalanceError("Insufficient balance")
  remaining_balance = balance - withdraw
  print(f"Withdraw successful. Remaining balance: {remaining_balance}")
except LowBalanceError as e:
  print(f"Error: {e}")
```

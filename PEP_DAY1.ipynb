```python
# Program 1: Take input, print its type, and double it (Datatypes)
user_input = input("Enter a value (number or text): ")
print(f"Type of input: {type(user_input)}")

try:
  # Try to convert to a number and double it
  numeric_value = float(user_input)
  print(f"Double of the input (numeric): {numeric_value * 2}")
except ValueError:
  # If not a number, treat as a string and concatenate
  print(f"Double of the input (string concatenation): {user_input * 2}")
```

```python
# find average of ages using sort
age = [19, 25, 36, 18, 23]

# Sort the age list
age.sort()
print(f"Sorted ages: {age}")

# Calculate the sum of ages
total_age = sum(age);

# Calculate the number of ages
num_ages = len(age);

# Calculate the average
if num_ages > 0:
  average_age = total_age / num_ages;
  print(f"The ages are: {age}");
  print(f"The average age is: {average_age}");
else:
  print("The list of ages is empty, cannot calculate average.");
```

```python
#make a list of 5 cities and run all the following operations: .append, .insert, .remove, .pop, sum(), max(), .reverse(), .sort(), len()

# Create a list of 5 cities
cities = ["Mumbai", "Delhi", "Bangalore", "Chennai", "Kolkata"]
print(f"Original cities list: {cities}")

# .append(): Add an element to the end
cities.append("Hyderabad")
print(f"After append('Hyderabad'): {cities}")

# .insert(): Add an element at a specific index
cities.insert(2, "Pune")
print(f"After insert(2, 'Pune'): {cities}")

# .remove(): Remove the first occurrence of a value
cities.remove("Mumbai")
print(f"After remove('Mumbai'): {cities}")

# .pop(): Remove and return an element at a given index (or last if no index)
popped_city = cities.pop(0) # Removes 'Delhi'
print(f"After pop(0): {cities}, Popped city: {popped_city}")

# sum(): Not applicable for list of strings. Would cause TypeError.
# max(): Find the lexicographically largest element
print(f"Max city (lexicographical): {max(cities)}")

# .reverse(): Reverse the order of elements in-place
cities.reverse()
print(f"After reverse(): {cities}")

# .sort(): Sort the elements in-place
cities.sort()
print(f"After sort(): {cities}")

# len(): Get the number of elements
print(f"Length of the list: {len(cities)}")
```

```python
#Tuple
days = ('Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday', 'Monday')
print(f"Original tuple: {days}")

# len() - length of the tuple
print(f"Length of the tuple: {len(days)}")

# Indexing - print element at index 1
print(f"Element at index 1: {days[1]}")

# .count() - count occurrences of 'Monday'
print(f"Count of 'Monday': {days.count('Monday')}")

# .index() - find index of 'Tuesday'
print(f"Index of 'Tuesday': {days.index('Tuesday')}")

# Slicing - print elements from index 1 up to (but not including) 2
print(f"Slice days[1:2]: {days[1:2]}")

# Membership check - 'monday' in days
print(f"Is 'monday' in days (case-sensitive): {'monday' in days}")
print(f"Is 'Monday' in days (case-sensitive): {'Monday' in days}")
```

```python
#make a tuple of 5 to 6 friends and perform these: len(), print([]), .count(), .index(), print(..[1:2]), print()

# Create a tuple of friends
friends = ("Srijan", "Parvez", "Nishit", "Ayush", "Prakash")
print(f"Original tuple of friends: {friends}")

# len(): Get the number of elements in the tuple
print(f"Length of the tuple: {len(friends)}")

# print([]): Interpreting this as printing an element by index
print(f"Friend at index 0: {friends[0]}")

# .count(): Count occurrences of an element
print(f"Count of 'Srijan': {friends.count('Srijan')}")

# .index(): Find the index of the first occurrence of an element
print(f"Index of 'Nishit': {friends.index('Nishit')}")

# print(..[1:2]): Slicing the tuple
print(f"Slice from index 1 to 2 (exclusive): {friends[1:2]}")

# print(): Print the entire tuple again
print(f"Tuple after operations: {friends}")
```

```python
#performing marks={.....}, print(marks), for i in marks: print (i), marks.add(100), marks.update([]), marks.remove(), marks.discend(), marks.pop(), print(a union (b)), print(a intersection(b)), print(c is subset (a)) on set of marks

# Initialize a set of marks
marks = {85, 92, 78, 65, 92}
print(f"Initial marks set: {marks}")

# Print elements using a for loop
print("Elements in marks set:")
for i in marks:
  print(i)

# marks.add(100)
marks.add(100)
print(f"After adding 100: {marks}")

# marks.update([]) - update with elements from another iterable
new_marks = {70, 85, 95}
marks.update(new_marks)
print(f"After updating with {new_marks}: {marks}")

# marks.remove() - remove an element (raises KeyError if not found)
try:
  marks.remove(65)
  print(f"After removing 65: {marks}")
except KeyError:
  print("65 not found in set for removal.")

# marks.discard() - remove an element if present, no error if not found
marks.discard(78)
print(f"After discarding 78: {marks}")
marks.discard(50) # Element not in set
print(f"After trying to discard 50 (not in set): {marks}")

# marks.pop() - remove and return an arbitrary element
if marks:
  popped_mark = marks.pop()
  print(f"Popped element: {popped_mark}, Set after pop: {marks}")
else:
  print("Set is empty, cannot pop element.")

# Demonstrate union, intersection, and subset with new sets
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
c = {1, 2}
d = {10, 11}

print(f"\nSet a: {a}")
print(f"Set b: {b}")
print(f"Set c: {c}")
print(f"Set d: {d}")

# print(a union (b))
print(f"Union of a and b: {a.union(b)}")

# print(a intersection(b))
print(f"Intersection of a and b: {a.intersection(b)}")

# print(c is subset (a))
print(f"Is c a subset of a?: {c.issubset(a)}")
print(f"Is a a subset of c?: {a.issubset(c)}")
print(f"Is d a subset of a?: {d.issubset(a)}")
```

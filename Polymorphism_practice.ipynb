## Polymorphism: Many Forms, One Interface

*   **Concept:** Polymorphism literally means "many forms." In programming, it refers to the ability of different objects to respond to the same method call (or function) in their own specific ways. It allows you to use a common interface for different data types or classes.

*   **Simple Idea:** Imagine you have a general command, but how that command is executed depends on *who* or *what* receives it. The same command behaves differently based on the object it's applied to.

### Real-World Example: "Make Sound"

Think about the command "**Make Sound**":

*   If you tell a **Dog** to "Make Sound", it will **bark**.
*   If you tell a **Cat** to "Make Sound", it will **meow**.
*   If you tell a **Cow** to "Make Sound", it will **moo**.

Here, "Make Sound" is the single interface (the common action), but the actual behavior (bark, meow, moo) is different for each animal. This is polymorphism in action!

```python
#POLYMORHISM
class Human:
  def speak(self):
    print("human speak")
class Baby:
  def speak(self):
    print("baby is crying")
class Dog:
  def speak(self):
    print("dog is barking")
#polymorphism in action
h=Human()
b=Baby()
d=Dog()
h.speak()
b.speak()
d.speak()
```

```python
#create class car, bike, private jet and each do polymorphism car is moving bike is moving and private jet is flying
class Car:
  def move(self):
    print("Car is moving on the road")

class SurferBoard:
  def move(self):
    print("Surfer board is surfing on sea tides")

class PrivateJet:
  def move(self):
    print("Private Jet is flying")

# Polymorphism in action
c = Car()
b = SurferBoard()
pj = PrivateJet()

c.move()
b.move()
pj.move()
```

```python
class Person:
  def role(self):
    print("I am a person")

class Student(Person):
  def role(self):
    print("I am a student")

class Teacher(Person):
  def role(self):
    print("I am a teacher")

class Assistant(Student, Teacher):
  pass # Inherits role() from its parent classes

# 1. Create object of Assistant
assistant = Assistant()

# 2. Call role()
assistant.role()

# 3. Observe which method is called (and optionally, inspect MRO)
print("\nMethod Resolution Order (MRO) for Assistant:")
print(Assistant.mro())
```

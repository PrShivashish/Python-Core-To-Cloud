```python
#List Q: Create list of numbers and print the elements
numbers = [1, 2, 3, 4, 5]

print("Elements of the list:")
for num in numbers:
    print(num)
```

```python
fruit = ["apple", "orange", "mango"] # Initialize fruit as a list of strings
print(fruit)
#append is used to add element in last
fruit.append("apple")
print(fruit)
```

```python
#Find the sum of all elements in a list
numbers=[5, 10, 15]
total = 0
for num in numbers:
    total += num
print("Sum of all elements:", total)
```

```python
#Reverse a list
nums = [1, 2, 3, 4]
reversed_list = nums[::-1]
print(reversed_list)
```

```python
#Remove duplicate element from thee list
nums = [1,2,2,3,4,4]
unique = []
for num in nums:
    if num not in unique:
        unique.append(num)
print(unique)
```

```python
#Tuple
colors=("red", "green", "blue")
print(colors[1])
print(colors[2])
```

```python
#conversion of list into a tuple
numbers = [1, 2, 3, 4, 5]
print(type(numbers))
numbers_tuple = tuple(numbers)
print(type(numbers_tuple))
```

```python
#conversion of list into a set
nums=[1,1,2,3,3,4]
nums_set=set(nums)
print(nums_set)
nums_list=list(nums_set)
print(nums_list)
```

```python
#conversion of dictionary into a set
statement = {"name":"shivashish","age":20}
student_set=set(statement)
print(type(student_set))
print(student_set)
```

```python
#conversion of set into a dictionary
subjects = {"math", "science", "history"}
subject_dict = dict.fromkeys(subjects)
print(subject_dict)
print(type(subject_dict))
```

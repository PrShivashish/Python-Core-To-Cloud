```python
#create a file named data, write name & city, append age, and read the file

# 1. Create a file named data.txt and write name & city
with open("/content/drive/MyDrive/data.txt", "w") as file:
  file.write("Name: Jane Doe\n")
  file.write("City: New York\n")

print("Initial content written to data.txt")

# 2. Append age to the file
with open("/content/drive/MyDrive/data.txt", "a") as file:
  file.write("Age: 30\n")

print("Age appended to data.txt")

# 3. Read the content of the file
with open("/content/drive/MyDrive/data.txt", "r") as file:
  data_content = file.read()
  print("\nContent of data.txt:")
  print(data_content)
```

```python
# Method 1: Iterating through the file line by line
line_count_method1 = 0
with open("/content/drive/MyDrive/students.txt", "r") as file:
  for line in file:
    line_count_method1 += 1

print(f"Number of lines (Method 1): {line_count_method1}")

# Method 2: Using readlines() and len()
with open("/content/drive/MyDrive/students.txt", "r") as file:
  lines = file.readlines()
  line_count_method2 = len(lines)

print(f"Number of lines (Method 2): {line_count_method2}")
```

```python
#library management system using oops concepts the system should calculate borrowing
#charges for different types of library items using oops concepts
# LibraryItem - Abstract class
# Book - child class
# Magazine - child class
# borrowing charge - encapsulated data
# LibraryApp - controller(HAS-A)
# output-
# Item Type: Book
# Borrow Days: 5
# Borrowing Charge: 50
# Item Type: Magazine
# Borrow Days: 3
# Borrowing Charge: 30

class LibraryItem:
    def __init__(self,item_id,title):
        self.item_id=item_id
        self.title=title
    def calculate_charge(self):
        pass

class Book(LibraryItem):
    def __init__(self,item_id,title,borrow_days):
        super().__init__(item_id,title)
        self._borrow_days=borrow_days
        #encapsulation because borrow days protected
    def calculate_charge(self):
        return self._borrow_days*10

class Magazine(LibraryItem):
    def __init__(self,item_id,title,borrow_days):
        super().__init__(item_id,title)
        self.borrow_days=borrow_days
    def calculate_charge(self):
        return self.borrow_days*10

class LibraryApp:
    def __init__(self):
        self.item=None
    def create_item(self,item_type):
        if item_type=="Book":
            self.item=Book(1,"Python Programming",5)
        else:
            self.item=Magazine(2,"Tech Magazine",3)
        return "Item Type: "+item_type
    def get_borrow_days(self):
        if hasattr(self.item,'_borrow_days'):
            return self.item._borrow_days
        else:
            return self.item.borrow_days
    def get_charge(self):
        return self.item.calculate_charge()

app=LibraryApp()
print(app.create_item("Book"))
print("Borrow Days:",app.get_borrow_days())
print("Borrowing Charge:",app.get_charge())
app2=LibraryApp()
print(app2.create_item("Magazine"))
print("Borrow Days:",app2.get_borrow_days())
print("Borrowing Charge:",app2.get_charge())
```

```python
#Implementation of 4 pillars:
# ABSTRACTION: LibraryItem acts as an abstract base class defining a blueprint
class LibraryItem:
    def __init__(self,item_id,title):
        self.item_id=item_id
        self.title=title

    # ABSTRACTION: Abstract method definition (to be implemented by subclasses)
    def calculate_charge(self):
        pass

# INHERITANCE: Book class inherits from LibraryItem
class Book(LibraryItem):
    def __init__(self,item_id,title,borrow_days):
        super().__init__(item_id,title) # INHERITANCE: Reusing parent constructor
        # ENCAPSULATION: _borrow_days is a protected member
        self._borrow_days=borrow_days

    # POLYMORPHISM: Overriding the parent method with Book-specific logic
    def calculate_charge(self):
        return self._borrow_days*10

# INHERITANCE: Magazine class inherits from LibraryItem
class Magazine(LibraryItem):
    def __init__(self,item_id,title,borrow_days):
        super().__init__(item_id,title) # INHERITANCE: Reusing parent constructor
        self.borrow_days=borrow_days

    # POLYMORPHISM: Overriding the parent method with Magazine-specific logic
    def calculate_charge(self):
        return self.borrow_days*10

class LibraryApp:
    def __init__(self):
        # HAS-A RELATIONSHIP (Composition): LibraryApp "has a" LibraryItem
        self.item=None

    def create_item(self,item_type):
        if item_type=="Book":
            self.item=Book(1,"Python Programming",5)
        else:
            self.item=Magazine(2,"Tech Magazine",3)
        return "Item Type: "+item_type

    def get_borrow_days(self):
        # ENCAPSULATION: Accessing data through a controlled class method
        if hasattr(self.item,'_borrow_days'):
            return self.item._borrow_days
        else:
            return self.item.borrow_days

    def get_charge(self):
        # POLYMORPHISM: The correct calculate_charge is called based on the object type
        return self.item.calculate_charge()

# Execution logic
app=LibraryApp()
print(app.create_item("Book"))
print("Borrow Days:",app.get_borrow_days())
print("Borrowing Charge:",app.get_charge())

app2=LibraryApp()
print(app2.create_item("Magazine"))
print("Borrow Days:",app2.get_borrow_days())
print("Borrowing Charge:",app2.get_charge())
```

```python
#Every line explained
#library management system using oops concepts the system should calculate borrowing
#charges for different types of library items using oops concepts
# LibraryItem - Abstract class
# Book - child class
# Magazine - child class
# borrowing charge - encapsulated data
# LibraryApp - controller(HAS-A)
# output-
# Item Type: Book
# Borrow Days: 5
# Borrowing Charge: 50
# Item Type: Magazine
# Borrow Days: 3
# Borrowing Charge: 30

class LibraryItem:  # parent class for all library items
    def __init__(self,item_id,title):  # constructor to initialize item details
        self.item_id=item_id  # Item ID
        self.title=title  # item title
    def calculate_charge(self):  # mthod to calcuate borrowing charge
        pass  # to be implemented by child classes


class Book(LibraryItem):  # child class inherting from libraryItem
    def __init__(self,item_id,title,borrow_days):
        super().__init__(item_id,title)  # calling parent class constrcutor
        self._borrow_days=borrow_days  # Protected attribute
        #encapsulation because borrow days protected
    def calculate_charge(self):  # overriding calculate_charge for books
        return self._borrow_days*10  # charge is 10 per day for books


class Magazine(LibraryItem):  # child class inheriting from LibrayItem
    def __init__(self,item_id,title,borrow_days):
        super().__init__(item_id,title)  # calling parent class constructor
        self.borrow_days=borrow_days  # public attribute for borrow days
    def calculate_charge(self):  # overriding calculate_charge for magazines
        return self.borrow_days*10  # charge is 10 per day for magazines


class LibraryApp:  # main controller class to manage library operation
    def __init__(self):
        self.item=None  # initially no item is selected
    def create_item(self,item_type):  # method to create library items
        if item_type=="Book":
            self.item=Book(1,"Python Programming",5)  # creating a Book object
        else:
            self.item=Magazine(2,"Tech Magazine",3)  # creating a magazine object
        return "Item Type: "+item_type
    def get_borrow_days(self):  # mthod to get borrow days from current item
        if hasattr(self.item,'_borrow_days'):  # check if item has protected attribute
            return self.item._borrow_days
        else:
            return self.item.borrow_days  # return public attribute
    def get_charge(self):  # Method to calculate charge for current item
        return self.item.calculate_charge()  # calls calculate_charge of the item object


app=LibraryApp()  # creating first instance of LibaryApp
print(app.create_item("Book"))  # crating a book item
print("Borrow Days:",app.get_borrow_days())  # getting borrow days for book
print("Borrowing Charge:",app.get_charge())  # calculating charge for book
app2=LibraryApp()  # creating second instance of libraryapp
print(app2.create_item("Magazine"))  # creating a magazine item
print("Borrow Days:",app2.get_borrow_days())  # getting borrow days for magazine
print("Borrowing Charge:",app2.get_charge())  # calculating charge for magazine
```

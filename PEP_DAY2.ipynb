```python
#now for mobile
mobile = {
    "brand": "samsung",
    "model": "S24",
    "price": 75000,
    "stock": 20
}

print("Original mobile dictionary:", mobile)

print("Values:", mobile.values())
print("Keys:", mobile.keys())
print("Items:", mobile.items())

mobile[24]=24
print("After adding mobile[24]=24:", mobile)

mobile.update({ "age":46, "course":"Django"})
print("After update for age and course:", mobile)

# Re-adding 'City' for pop demonstration, similar to Student dictionary
mobile['City'] = 'Delhi'
print("After adding 'City':", mobile)

mobile.pop("City")
print("After popping 'City':", mobile)

mobile.popitem()
print("After popitem():", mobile)

d2_mobile = mobile.copy()
print("Copied dictionary (d2_mobile):", d2_mobile)

print("Length of mobile dictionary:", len(mobile))
```

```python
contact={}
while True:
  print("\n----contact book-----")
  print("1. Add contact")
  print("2. view contact")
  print("3. search contact")
  print("4. delete contact")
  print("5. exit")
  choice=int(input("enter your choice:"))

  #for adding the contacts
  if choice==1:
    name=input("Enter name: ")
    phone=input("Enter phone number: ")
    contact[name] = phone
    print("Contact added successfully!")

  #for view contact
  elif choice == 2:
    if contact:
      print("\nsaved contact: ")
      for name, phone in contact.items():
        print(f"Name: {name}, Phone: {phone}")
    else:
      print("No contacts found.")

  #for search contact
  elif choice == 3:
    name = input("Enter name to search: ")
    if name in contact:
      print(f"Name: {name}, Phone: {contact[name]}")
    else:
      print("Contact not found.")

  #delete contact
  elif choice == 4:
    name = input("Enter name to delete: ")
    if name in contact:
      del contact[name]
      print("Contact deleted successfully!")
    else:
      print("Contact not found.")

  #exit
  elif choice == 5:
    print("Thank you for using contact book")
    break

  else:
    print("Invalid choice. Please select a valid option.")
```

```python
#palindrome
def is_palindrome(s):
    # Remove spaces and convert to lowercase for accurate comparison
    s = s.replace(" ", "").lower()
    return s == s[::-1]

# Test cases
string1 = "madam"
string2 = "A man a plan a canal Panama"
string3 = "hello"

print(f"'{string1}' is palindrome: {is_palindrome(string1)}")
print(f"'{string2}' is palindrome: {is_palindrome(string2)}")
print(f"'{string3}' is palindrome: {is_palindrome(string3)}")
```

```python
#palindrome using pointers
def is_palindrome_two_pointers(s):
    # Remove spaces and convert to lowercase for accurate comparison
    s = s.replace(" ", "").lower()

    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

# Test cases for two-pointer method
string1_tp = "madam"
string2_tp = "A man a plan a canal Panama"
string3_tp = "hello"

print(f"'{string1_tp}' is palindrome (two-pointers): {is_palindrome_two_pointers(string1_tp)}")
print(f"'{string2_tp}' is palindrome (two-pointers): {is_palindrome_two_pointers(string2_tp)}")
print(f"'{string3_tp}' is palindrome (two-pointers): {is_palindrome_two_pointers(string3_tp)}")
```

```python
#Count character frequency of a string
def count_char_frequency(s):
    frequency = {}
    for char in s:
        frequency[char] = frequency.get(char, 0) + 1
    return frequency

# Test case
text = "python programming"
char_counts = count_char_frequency(text)
print(f"Character frequencies in '{text}': {char_counts}")

text2 = "hello world"
char_counts2 = count_char_frequency(text2)
print(f"Character frequencies in '{text2}': {char_counts2}")
```

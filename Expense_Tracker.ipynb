```python
# start
# load old expenses from the file
# show menu
# 1. Add expense
# 2. Show expense
# 3. Total expense
# 4. exit
# user selects one option
# function runs
# save data
# repeat

file_name = "expenses.txt"
expenses = []

print(f"Expense data file name initialized to: {file_name}")
print(f"Expenses list initialized as: {expenses}")


#load expense
def load_expenses():
    global expenses
    try:
        with open(file_name, "r") as file:
            expenses.clear() # Clear existing data to avoid duplicates on reload
            for line in file:
                try:
                    amount_str, category = line.strip().split(',', 1)
                    amount = float(amount_str)
                    expenses.append({"amount": amount, "category": category})
                except ValueError:
                    print(f"Skipping malformed expense line: {line.strip()}")
    except FileNotFoundError:
        print("No previous expense data found. Starting with an empty list.")
        expenses = [] # Ensure expenses is empty if file not found
    except Exception as e:
        print(f"An error occurred while loading expenses: {e}")

print("Defined load_expenses() function.")


#save expenses
def save_expenses():
    global expenses
    try:
        with open(file_name, "w") as file:
            for expense in expenses:
                file.write(f"{expense['amount']},{expense['category']}\n")
        print("Expenses saved successfully.")
    except Exception as e:
        print(f"An error occurred while saving expenses: {e}")

print("Defined save_expenses() function.")

#add expense
def add_expense():
    global expenses
    try:
        amount = float(input("Enter expense amount: "))
        category = input("Enter expense category: ")
        expenses.append({"amount": amount, "category": category})
        save_expenses()
        print("Expense added successfully.")
    except ValueError:
        print("Invalid amount. Please enter numbers only.")
    except Exception as e:
        print(f"An error occurred while adding expense: {e}")

print("Defined add_expense() function.")

#show expense
def show_expenses():
    global expenses
    if not expenses:
        print("No expenses recorded yet.")
    else:
        print("\n--- Your Expenses ---")
        for i, expense in enumerate(expenses):
            print(f"{i + 1}. Amount: {expense['amount']:.2f}, Category: {expense['category']}")
        print("---------------------")

print("Defined show_expenses() function.")


#calculate total expense
def total_expense():
    global expenses
    if not expenses:
        print("No expenses recorded yet to calculate total.")
        return

    total = 0
    for expense in expenses:
        total += expense['amount']

    print(f"\nTotal Expenses: {total:.2f}")

print("Defined total_expense() function.")



def main():
    load_expenses() # Load data when program starts

    while True:
        print("\n----- Expense Tracker Menu -----")
        print("1. Add expense")
        print("2. Show expenses")
        print("3. Total expense")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_expense()
        elif choice == "2":
            show_expenses()
        elif choice == "3":
            total_expense()
        elif choice == "4":
            print("Thank you for using the Expense Tracker!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

# Call the main function to start the program
main()
```
